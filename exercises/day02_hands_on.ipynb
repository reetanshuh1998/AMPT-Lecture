## Problem 4: Mass-Dependent Mid-rapidity Dip in $\mathrm{d}N/\mathrm{d}\eta$
👉 **[INTERACTIVE SIMULATOR]:** Open the simulator **[Phase-Space Jacobian Explorer](../day2/day2_rapidity_jacobian_explorer.html)** and sweep the particle mass and transverse momentum sliders. Notice how the phase space density suppression at mid-rapidity develops for heavy particles at low $p_T$!


The pseudorapidity distribution $\mathrm{d}N/\mathrm{d}\eta$ is related to the rapidity distribution $\mathrm{d}N/\mathrm{d}y$ by Sahoo Eq. 5.147:
$$\frac{\mathrm{d}N}{\mathrm{d}\eta \mathrm{d}p_T} = J(y \to \eta) \frac{\mathrm{d}N}{\mathrm{d}y \mathrm{d}p_T} = \sqrt{1 - \frac{m_0^2}{m_T^2 \cosh^2 y}} \frac{\mathrm{d}N}{\mathrm{d}y \mathrm{d}p_T}$$

At mid-rapidity ($y = 0 \implies \eta = 0$), the Jacobian becomes:
$$J(0) = \frac{p_T}{m_T} = \frac{p_T}{\sqrt{p_T^2 + m_0^2}}$$

This causes a non-physical dip at $\eta = 0$ for massive particles (kaons and protons) while leaving light particles (pions) unaffected.

$$c_s^2 = \frac{-4\ln(\sqrt{s_{NN}}/2m_p)}{3\sigma_y^2} + \sqrt{\left(\frac{4\ln(\sqrt{s_{NN}}/2m_p)}{3\sigma_y^2}\right)^2 + 1}$$
